3. Which Category has largest number of installations ?


```python
df_cat_inst= df_copy.groupby(['Category'])['Installs'].sum().sort_values(ascending=False).reset_index()
df_cat_inst.Installs= df_cat_inst.Installs/1000000000
dfhi=df_cat_inst.head(10)
plt.figure(figsize=(10,5))
sns.set_context('talk')
sns.set_style('darkgrid')
ax= sns.barplot(x='Installs',y='Category',data=dfhi,palette='husl')
ax.set_xlabel("No. of Installations in Billion")
ax.set_ylabel('')
ax.set_title("Most Popular Categories in Play Store", size = 20)
```


what are the top 5 most installed app in GAME', 'COMMUNICATION', 'PRODUCTIVITY', 'SOCIAL these  categories ?

```python
dfa= df_copy.groupby(['App','Category'])['Installs'].sum().sort_values(ascending=False).reset_index()
dfa.rename(columns={'Installs':'Install in millions'},inplace=True)
apps=['GAME', 'COMMUNICATION', 'PRODUCTIVITY', 'SOCIAL']
sns.set_context('poster')
sns.set_style('darkgrid')
plt.figure(figsize=(40,30))
for i, app in enumerate(apps):
    df2=dfa[dfa.Category == app]
    df3=df2.head(5)
    plt.subplot(4,2, i+1)
    sns.barplot(data=df3,x='Install in millions',y='App',palette='husl')
    plt.ylabel('')
    plt.title(app,size=20)
plt.tight_layout()
plt.subplots_adjust()
plt.show()
```
